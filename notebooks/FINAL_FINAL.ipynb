```python
import yaml

with open("/content/MIA_FINAL_CV/data.yaml", "r") as f:
    data = yaml.safe_load(f)

print(data)
```

```python
from ultralytics import YOLO

# Start from pretrained YOLO11n
model = YOLO("yolo11n.pt")

results = model.train(
    data="/content/MIA_FINAL_CV/data.yaml",

    # Training
    epochs=80,
    imgsz=640,
    batch=16,
    device=0,

    # Scale / Zoom
    scale=0.5,

    # Position
    translate=0.10,

    # Rotation
    degrees=10,

    # Perspective
    perspective=0.0005,

    # Lighting / Color
    hsv_h=0.015,
    hsv_s=0.4,
    hsv_v=0.4,

    # Flip
    fliplr=0.5,
    flipud=0.0,

    # Mosaic
    mosaic=1.0,

    # Training output
    project="/content/runs",
    name="scroll_detection_v2",
    exist_ok=True
)
```

```python
from ultralytics import YOLO

model = YOLO(
    "/content/runs/scroll_detection_v2/weights/best.pt"
)

results = model.val(
    data="/content/MIA_FINAL_CV/data.yaml",
    split="test",
    plots=True
)

print("Precision :", results.box.mp)
print("Recall    :", results.box.mr)
print("mAP50     :", results.box.map50)
print("mAP50-95  :", results.box.map)
```
